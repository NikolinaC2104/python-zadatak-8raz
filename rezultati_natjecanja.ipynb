## Zadatak

Na školskom natjecanju iz informatike sudjelovalo je pet učenika.

Napiši program koji će:

1. omogućiti unos broja bodova za pet učenika
2. spremiti unesene bodove u listu
3. sortirati bodove od najvećeg prema najmanjem
4. ispisati sortirane rezultate
5. ispisati najbolji ostvareni rezultat

** Napomena:** Za sortiranje nemoj koristiti 'sort()' ili 'sorted()'. Potrebno je primijeniti algoritam sortiranja uspoređivanjem i zamjenom elemenata.

## 1. korak - Stvaranje liste

Bodove učenika spremamo u listu.
Napravimo listu 'bodovi' koja će sadržavati pet početnih vrijednosti '0'.

```python
bodovi = [0, 0, 0, 0, 0]
print(bodovi)
```

## 2. korak - Unos bodova

Sada trebamo unijeti bodove za pet učenika.
Budući da isti postupak trebamo ponoviti pet puta, koristiti ćemo petlju 'for'.
Pomoću 'input()' unosimo podatak, a pomoću 'int()' pretvaramo unesenu vrijednost u cijeli broj.
Vrijednost spremamo na odgovarajuće mjesto u listi 'bodovi'.

```python
for i in range(5):
    bodovi[i] = int(input("Unesi bodove " + str(i + 1) + ". učenika: "))
```

### Zašto koristimo 'str(i+1)'?

To nam odgovara za pristup elementima liste jer indeksi liste počinju od '0'. Međutim, pri ispisu želimo učenike označiti brojevima od 1 do 5, pa koristimo 'i+1'.
Funkcija 'str()' pretvara broj u tekst kako bismo ga mogli spojiti s ostalim tekstom pomoću operatora '+'.
Npr. kada je 'i=0':
- i+1 daje broj 1
- str(i+1) daje tekst 1
Zato će program ispisati:
- Unesi bodove 1. učenika:

### Provjera unesenih podataka

```python
print(bodovi)
```

## 3. korak - Zamjena elemenata

Bodove trebamo poredati od najvećeg prema najmanje.
Uspoređivati ćemo dva susjedna elementa liste. Ako je prvi element manji od sljedećeg, zamijeniti ćemo im mjesta.
Za zamjenu ćemo koristi pomoćnu varijablu.

```python
 a = 72
 b = 88

 if a<b:
     pomocna = a
     a = b
     b = pomocna

print(a,b)
```

### Razmisli

2. Čemu služi varijabla 'pomocna'?

### 4. korak - Sortiranje cijele liste

Sada ćemo isti postupak uspoređivanja i zamjene primijeniti na cijelu listu.
Uspoređujemo dva susjedna elementa liste. Ako je element s lijeve strane manji od elementa s desne strane, zamijeniti ćemo im mjesta.
Postupak ponavljamo više puta kako bi svi bodovi bili poredani od najvećeg prema najmanjem.
Za to su nam potrebne dvije petlje:
- vanjska petlja koja ponavlja postupak sortiranja
- unutarnja petlja koja prolazi kroz susjedne elemente liste i uspoređuje ih

```python
for i in range(5):
    for j in range(4):
        if bodovi[j] < bodovi[j+1]:
            pomocna = bodovi[j]
            bodovi[j] = bodovi[j+1]
            bodovi[j+1] = pomocna
```

### Provjera sortiranja

Ispišimo listu nakon sortiranja i provjerimo jesu li rezultati poredani od najvećeg prema najmanjem.

```python
print(bodovi)
```

### Kako radi algoritam?

U algoritmu koristimo dvije petlje.
for i in range(5):
to je vanjska petlja koja ponavlja postupak sortiranja više puta
for j in range(4):
to je unutarnja petlja koja prolazi kroz listu i omogućuje uspoređivanje susjednih elemenata. Varijabla j predstavlja indeks elementa koji trenutno promatramo.
Uspoređujemo bodovi[j] i bodovi[j+1] , npr. kada je j=0 onda uspoređujemo bodovi[0] i bodovi[1], kada je j=1 uspoređujemo bodovi[1]i bodovi[2] i tako dalje. Ako je lijevi element manji od desnog, zamjenjujemo im mjesta.

Zašto vanjska petlja koristi range(4) a ne range(5)?
- zato što lista ima pet elemenata s indeksima od 0 do 4, budući da uspoređujemo bodovi[j] s bodovi[j+1], najveća vrijednost varijable je smije biti 3 te tada posljednji put uspoređujemo bodovi[3] i bodovi[4].

## 5. korak - Ispis rezultata

Nakon sortiranja ispisat ćemo sve rezultate od najvećeg prema najmanjem.
Za ispis svih elemenata liste ponovno koristimo petlju 'for'.

```python
print("Rezultati od najvećeg prema najmanjem:")

for i in range(5):
    print(bodovi[i])
```

## 6. korak - Ispis najboljeg rezultata

Budući da smo bodove sortirali od najvećeg prema najmanje, najbolji rezultat se nalazi na prvom mjestu liste, odnosno on ima indeks 0

```python
print("Najbolji rezultat je", bodovi[0], ".")
```

## Cjelo rješenje

U nastavku se nalazi cijeli program povezan u cjelinu.

Program radi:
- omogućuje unos bodova za pet učenika
- sprema bodove u listu
- sortira rezultate od najvećeg prema najmanjem
- ispisuje sortirane rezultate
- ispisuje najbolji rezultat

```python
# početni rezultati natjecanja - lista
bodovi = [0 ,0, 0, 0, 0]

# unos bodova za svakog učenika
for i in range(5):
    bodovi[i] = int(input("Unesi bodove " + str(i+1) + ". učenika: "))

# sortiranje od najvećeg prema najmanjem
for i in range(5):
    for j in range(4):
        if bodovi[j] < bodovi[j+1]:
            pomocna = bodovi[j]
            bodovi[j] = bodovi [j+1]
            bodovi[j+1]= pomocna

# ispis rezultata
print()
print("Rezultati od najvećeg prema najmanjem:")

for i in range(5):
    print(bodovi[i])

# ispis najboljeg rezultata
print()
print("Najbolji rezultat je", bodovi[0], ".")
```

### Zašto koristimo 'print()' bez sadržaja?

Naredba 'print()' bez sadržaja ispisuje prazan red. U ovom programu ju koristimo kako bi ispis rezultata bio pregledniji i kako bismo vizualno odvojili različite dijelove ispisa.
Ova naredba nije potrebna za rad programa i njezino uklanjanje ne bi utjecalo na rezultat programa, već samo na izgled ispisa.
