### Zašto koristimo `str(i + 1)`?

Petlja `for i in range(5)` daje vrijednosti varijable `i` od `0` do `4`.

To odgovara indeksima liste. Međutim, pri ispisu želimo učenike označiti brojevima od 1 do 5, pa koristimo `i + 1`.

Funkcija `str()` pretvara broj u tekst kako bismo ga mogli spojiti s ostalim tekstom pomoću operatora `+`.

### Rješenje 3. koraka:

```python
for i in range(5):
    for j in range(4):
        if bodovi[j] < bodovi[j + 1]:
            pomocna = bodovi[j]
            bodovi[j] = bodovi[j + 1]
            bodovi[j + 1] = pomocna
```

### Kako radi algoritam?
U algoritmu koristimo dvije petlje. for i in range(5): to je vanjska petlja koja ponavlja postupak sortiranja više puta for j in range(4): to je unutarnja petlja koja prolazi kroz listu i omogućuje uspoređivanje susjednih elemenata. Varijabla j predstavlja indeks elementa koji trenutno promatramo. Uspoređujemo bodovi[j] i bodovi[j+1] , npr. kada je j=0 onda uspoređujemo bodovi[0] i bodovi[1], kada je j=1 uspoređujemo bodovi[1]i bodovi[2] i tako dalje. Ako je lijevi element manji od desnog, zamjenjujemo im mjesta.

Zašto vanjska petlja koristi range(4) a ne range(5)?

zato što lista ima pet elemenata s indeksima od 0 do 4, budući da uspoređujemo bodovi[j] s bodovi[j+1], najveća vrijednost varijable je smije biti 3 te tada posljednji put uspoređujemo bodovi[3] i bodovi[4].

### Rješenje 4. koraka:

```python
print("Rezultati od najvećeg prema najmanjem:")

for i in range(5):
    print(bodovi[i])
```

```python
print("Najbolji rezultat je", bodovi[0], "bod.")
```

## Cjelo rješenje

U nastavku se nalazi cijeli program povezan u cjelinu.



```python
# Rezultati školskog natjecanja

bodovi = [0, 0, 0, 0, 0]

# Unos bodova
for i in range(5):
    bodovi[i] = int(input("Unesi bodove " + str(i + 1) + ". učenika: "))

# Sortiranje od najvećeg prema najmanjem
for i in range(5):
    for j in range(4):
        if bodovi[j] < bodovi[j + 1]:
            pomocna = bodovi[j]
            bodovi[j] = bodovi[j + 1]
            bodovi[j + 1] = pomocna

# Ispis rezultata
print()
print("Rezultati od najvećeg prema najmanjem:")

for i in range(5):
    print(bodovi[i])

# Ispis najboljeg rezultata
print()
print("Najbolji rezultat je", bodovi[0], "bod.")
```

### Zašto koristimo 'print()' bez sadržaja?

Naredba 'print()' bez sadržaja ispisuje prazan red. U ovom programu ju koristimo kako bi ispis rezultata bio pregledniji i kako bismo vizualno odvojili različite dijelove ispisa.
Ova naredba nije potrebna za rad programa i njezino uklanjanje ne bi utjecalo na rezultat programa, već samo na izgled ispisa.
